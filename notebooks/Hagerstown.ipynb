### Gathering team info in dictionary

```python
stats = {}
for year in years:
  league = League(league_id=league_id, year=year, swid=swid, espn_s2=espn_s2)
  for team in league.teams:
    full_name = team.owners[0]['firstName'] + ' ' + team.owners[0]['lastName']
    #print(year, team.team_name, "finished in position", team.final_standing, "and", league.settings.playoff_team_count, "made the playoffs")
    if full_name not in stats.keys():
      d = {}
      d['name'] = full_name
      d['wins'] = team.wins
      d['losses'] = team.losses

      if team.final_standing != 0: # if the season is over
        d['playoff_apps'] = int(team.final_standing <= league.settings.playoff_team_count)
      elif league.current_week > league.settings.reg_season_count: # if season is ongoing but playoffs have started
        d['playoff_apps'] = int(team.standing <= league.settings.playoff_team_count)
      else: # if playoffs have not been determined yet
        d['playoff_apps'] = 0

      d['seasons'] = 1
      stats[full_name] = d
    else:
      stats[full_name]['wins'] += team.wins
      stats[full_name]['losses'] += team.losses

      if team.final_standing != 0: # if the season is over
        stats[full_name]['playoff_apps'] += int(team.final_standing <= league.settings.playoff_team_count)
      elif league.current_week > league.settings.reg_season_count: # if season is ongoing but playoffs have started
        stats[full_name]['playoff_apps'] += int(team.standing <= league.settings.playoff_team_count)

      stats[full_name]['seasons'] += 1

print(stats)

values = [v for v in stats.values()]
```

# Waiver Studs

```python
def player_table(year: int):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  # creating df for player_map
  player_id_to_name = {k: v for k, v in league.player_map.items() if isinstance(k, int)}
  player_df = pd.DataFrame.from_dict(player_id_to_name, orient='index', columns=['Player Name']).reset_index()
  player_df.columns = ['Player ID', 'Player Name']

  # gathering scores from players while they were in lineups
  scores = defaultdict(list)
  positions = defaultdict(str)

  for i in range(1, league.current_week + 1):
    scoreboard = league.box_scores(i)
    for matchup in scoreboard:
      for player in matchup.home_lineup:
        if player.lineupSlot not in ['BE', 'IR']:
          scores[player.playerId].append(player.points)
          positions[player.playerId] = player.position
      for player in matchup.away_lineup:
        if player.lineupSlot not in ['BE', 'IR']:
          scores[player.playerId].append(player.points)
          positions[player.playerId] = player.position

  # adding positions to player_df
  positions_df = pd.DataFrame.from_dict(positions, orient='index', columns=['Position']).reset_index()
  positions_df.columns = ['Player ID', 'Position']

  player_df = player_df.merge(positions_df, on='Player ID')

  return player_df, scores
```

```python
def waiver_points(year: int, scores):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  # creating list of playerId that were drafted
  draft = league.draft
  drafted = []
  for i in range(len(draft)):
    drafted.append(draft[i].playerId)

  # removing drafted players from scores
  undrafted_scores = dict(scores)
  for id in drafted:
    if id in undrafted_scores.keys():
      del undrafted_scores[id]

  # calculating point sums for all free agent players
  total_points = defaultdict(list)
  for k, v in undrafted_scores.items():
    total_points[k].append(sum(v))
    total_points[k].append(len(v))

  return total_points
```

```python
def replacements(year: int):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  # initialize dictionary and weekly ranks of replacement players
  rep = defaultdict(list)

  # loop through each lineup and append starters' points to week_scores dictionary
  for i in range(1, league.current_week + 1):
    week_scores = defaultdict(list)
    scoreboard = league.box_scores(i)

    for matchup in scoreboard:
      for player in matchup.home_lineup + matchup.away_lineup:
        if player.lineupSlot not in ['BE', 'IR'] and player.points != 0 and player.position in ['QB', 'RB', 'WR', 'TE']:
          week_scores[player.position].append(player.points)

    # sort weekly scores and take average of bottom 3 starters to rep dictionary
    for key in week_scores.keys():
      sorted_scores = sorted(week_scores[key], reverse=True)
      bot_3 = sorted_scores[-3:]
      rep[key].append(mean(bot_3))

  return rep
```

```python
def top_waiver_adds(year: int):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  # gather player_df, waiver_points, replacements
  player_df, scores = player_table(year)
  total_points = waiver_points(year, scores)
  rep = replacements(year)

  # viewing top waiver adds
  total_points_df = pd.DataFrame.from_dict(total_points, orient='index', columns=['Total Points', 'Weeks']).reset_index()
  total_points_df.columns = ['Player ID', 'Total Points', 'Weeks']
  waiver_adds_df = total_points_df.merge(player_df, on='Player ID')
  waiver_adds_df['Year'] = year

  # positional standards

  pos_standards = {
      "QB": mean(rep['QB']),
      "RB": mean(rep['RB']),
      "WR": mean(rep['WR']),
      "TE": mean(rep['TE'])
  }

  print(year, ':', pos_standards)

  # calculating VOR
  waiver_adds_df['VOR'] = round(waiver_adds_df['Total Points'] - (waiver_adds_df['Weeks'] * waiver_adds_df['Position'].map(pos_standards)), 2)
  waiver_adds_df = waiver_adds_df.sort_values(by=['VOR'], ascending=False).reset_index(drop=True)

  return waiver_adds_df
```

```python
all_waivers_df = pd.DataFrame()
for year in years:
  all_waivers_df = pd.concat([all_waivers_df, top_waiver_adds(year)])
all_waivers_df = all_waivers_df.sort_values(by=['VOR'], ascending=False).reset_index(drop=True)
all_waivers_df.head(10)
```

```python
def reg_scores_dict(year: int):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  reg = league.settings.reg_season_count

  # grab list of scores from each team that season
  scores = {}
  for team in league.teams:
    full_name = team.owners[0]['firstName'] + ' ' + team.owners[0]['lastName']
    scores[full_name] = team.scores[:reg]

  return scores
```

```python
def playoff_scores_dict(year: int):

  # initialize league
  league = League(
      league_id=league_id,
      year=year,
      swid=swid,
      espn_s2=espn_s2)

  reg = league.settings.reg_season_count

  # grab all scores from the first playoff week (when everyone has something to play for)
  scores = defaultdict(list)
  for box in league.box_scores(reg+1):
    if box.home_team != 0 and box.away_team != 0:
      # add home team score to dict
      home_name = box.home_team.owners[0]['firstName'] + ' ' + box.home_team.owners[0]['lastName']
      scores[home_name].append(box.home_score)
      # add away team score to dict
      away_name = box.away_team.owners[0]['firstName'] + ' ' + box.away_team.owners[0]['lastName']
      scores[away_name].append(box.away_score)

  # grab all subsequent scores from top 2 finishers (people with motivations all the way through)
  for wk in range(reg+2, 18):
    for box in league.box_scores(wk):
      if box.home_team != 0 and box.away_team != 0:
        # add home team score to dict
        if box.home_team.final_standing <= 2:
          home_name = box.home_team.owners[0]['firstName'] + ' ' + box.home_team.owners[0]['lastName']
          score = box.home_score
          if wk >= 16:
            score /= 2
          scores[home_name].append(score)
        # add away team score to dict
        if box.away_team.final_standing <= 2:
          away_name = box.away_team.owners[0]['firstName'] + ' ' + box.away_team.owners[0]['lastName']
          score = box.away_score
          if wk >= 16:
            score /= 2
          scores[away_name].append(score)

  return scores
```
